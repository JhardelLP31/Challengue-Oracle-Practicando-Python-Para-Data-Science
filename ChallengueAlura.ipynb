### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Sumamos el total de ventas de cada tienda
fact_1 = tienda['Precio'].sum()
fact_2 = tienda2['Precio'].sum()
fact_3 = tienda3['Precio'].sum()
fact_4 = tienda4['Precio'].sum()

print(f"Facturación Tienda 1: ${fact_1:,.2f}")
print(f"Facturación Tienda 2: ${fact_2:,.2f}")
print(f"Facturación Tienda 3: ${fact_3:,.2f}")
print(f"Facturación Tienda 4: ${fact_4:,.2f}")
```

Interpretación: Al sumar todo, me di cuenta de que la Tienda 1 es la que más plata está moviendo, superando los 1,150 millones. Por otro lado, la Tienda 4 es la que menos vende de todas. Hay una diferencia de más de 100 millones entre ellas, lo que ya nos da una pista de qué tienda podría ser la menos rentable para el señor Juan.

# 2. Ventas por categoría

```python
print("Top categorías Tienda 1:\n", tienda['Categoría del Producto'].value_counts().head(3))
print("\nTop categorías Tienda 2:\n", tienda2['Categoría del Producto'].value_counts().head(3))
print("\nTop categorías Tienda 3:\n", tienda3['Categoría del Producto'].value_counts().head(3))
print("\nTop categorías Tienda 4:\n", tienda4['Categoría del Producto'].value_counts().head(3))
```

Interpretación: Lo que más se vende en todas las tiendas son Muebles y Electrónicos. Parece que a la gente le gusta comprar cosas grandes en Alura Store. Juguetes queda en tercer lugar en todas, así que el fuerte del negocio definitivamente no son los niños, sino el hogar y la tecnología.

# 3. Calificación promedio de la tienda


```python
prom_1 = round(tienda['Calificación'].mean(), 2)
prom_2 = round(tienda2['Calificación'].mean(), 2)
prom_3 = round(tienda3['Calificación'].mean(), 2)
prom_4 = round(tienda4['Calificación'].mean(), 2)

print(f"Calificación Promedio Tienda 1: {prom_1}")
print(f"Calificación Promedio Tienda 2: {prom_2}")
print(f"Calificación Promedio Tienda 3: {prom_3}")
print(f"Calificación Promedio Tienda 4: {prom_4}")
```

Interpretación: Aquí los números están muy parejos, todos rondan el 4.0. Pero si nos ponemos detallistas, la Tienda 1 es la única que bajó del 4 (tiene 3.98). No es una mala nota, pero comparada con las otras, parece que los clientes de la Tienda 1 están un poquito menos conformes.

# 4. Productos más y menos vendidos

```python
# Tienda 1
print(f"Tienda 1 - Más vendido: {tienda['Producto'].value_counts().idxmax()}")
print(f"Tienda 1 - Menos vendido: {tienda['Producto'].value_counts().idxmin()}")

# Tienda 2
print(f"Tienda 2 - Más vendido: {tienda2['Producto'].value_counts().idxmax()}")
print(f"Tienda 2 - Menos vendido: {tienda2['Producto'].value_counts().idxmin()}")

# Tienda 3
print(f"Tienda 3 - Más vendido: {tienda3['Producto'].value_counts().idxmax()}")
print(f"Tienda 3 - Menos vendido: {tienda3['Producto'].value_counts().idxmin()}")

# Tienda 4
print(f"Tienda 4 - Más vendido: {tienda4['Producto'].value_counts().idxmax()}")
print(f"Tienda 4 - Menos vendido: {tienda4['Producto'].value_counts().idxmin()}")
```

Interpretación: Aquí hay variedad. Mientras que en la Tienda 1 lo que manda son los Microondas, en la Tienda 4 el producto estrella es la Cama box. También es interesante ver que la "Guitarra eléctrica" es lo que menos se vende en la Tienda 4; parece que el público de esa zona no es muy musical o prefiere otros instrumentos.

# 5. Envío promedio por tienda

```python
# Calculamos el promedio de la columna 'Costo de envío' y redondeamos a 2 decimales
envio_1 = round(tienda['Costo de envío'].mean(), 2)
envio_2 = round(tienda2['Costo de envío'].mean(), 2)
envio_3 = round(tienda3['Costo de envío'].mean(), 2)
envio_4 = round(tienda4['Costo de envío'].mean(), 2)

print(f"Costo de envío promedio Tienda 1: ${envio_1}")
print(f"Costo de envío promedio Tienda 2: ${envio_2}")
print(f"Costo de envío promedio Tienda 3: ${envio_3}")
print(f"Costo de envío promedio Tienda 4: ${envio_4}")
```

Interpretación: Los costos de envío son altos en todas, pero la diferencia es clara. La Tienda 1 es la más costosa en logística, mientras que la Tienda 4 es la más eficiente, costando unos $2,500 menos en promedio. Sin embargo, hay que ver si este ahorro compensa su baja facturación.

Es verdad que esta tienda tiene buenos costos de envío y buenas calificaciones, pero al final del día es la que menos dinero hace entrar a la empresa. Hay una diferencia de más de 100 millones entre lo que vende la Tienda 1 y la Tienda 4.

Aunque la Tienda 1 gasta mucho en envíos, es la que más vende, así que no conviene soltarla. Lo mejor para juntar capital es vender la sucursal que menos factura, que en este caso es la 4.
